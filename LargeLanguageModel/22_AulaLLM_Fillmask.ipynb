```markdown
### O que este passo faz e por que é necessário?
Este bloco importa a biblioteca inteira `transformers` e, em seguida, importa especificamente a função `pipeline`. O `pipeline` é o ponto de partida mais fácil para iniciantes: ele esconde toda a complexidade de converter palavras em números (tokenização) e realizar cálculos matemáticos complexos, entregando uma interface simples para usarmos a IA de forma direta.

---

## Passo 3: Criando o Pipeline e Preenchendo uma Máscara Simples
Agora que temos nossas ferramentas prontas, vamos carregar o modelo de linguagem em português e testá-lo com um desafio simples: adivinhar uma palavra oculta representada pela etiqueta `[MASK]`.
```

````python
mascarar = pipeline("fill-mask", model="neuralmind/bert-base-portuguese-cased")
texto = mascarar("Existe uma chance do corpo cair no [MASK]")
for x in range(len(texto)):
  print(texto[x])
````

```markdown
### O que este passo faz?
1. Criamos um objeto chamado `mascarar` usando `pipeline("fill-mask", model="neuralmind/bert-base-portuguese-cased")`. Isso faz o download do modelo BERTimbau (caso ainda não tenha sido baixado) e o configura especificamente para a tarefa de preencher lacunas (*fill-mask*).
2. Chamamos `mascarar("Existe uma chance do corpo cair no [MASK]")`. A palavra especial `[MASK]` indica ao modelo onde está o "buraco" que ele deve preencher.
3. Iteramos sobre a lista de respostas sugeridas pelo modelo e as exibimos na tela.

### Por que foi escrito assim?
O BERT (Bidirectional Encoder Representations from Transformers) foi originalmente treinado ocultando palavras aleatórias de um texto e forçando o modelo a adivinhar quais eram elas com base no contexto ao redor (as palavras antes e depois da máscara). Por isso, escrever a consulta com `[MASK]` faz uso direto da habilidade nativa do modelo.

### Como interpretar o resultado?
O modelo retorna uma lista de dicionários (geralmente as 5 melhores opções por padrão). Cada opção contém:
* `'score'`: A probabilidade ou nível de certeza que o modelo tem sobre aquela palavra (varia de 0 a 1, onde 1 é 100% de certeza).
* `'token_str'`: A palavra sugerida para preencher a lacuna.
* `'sequence'`: A frase completa reconstruída com a sugestão aplicada.

Note como a palavra **"chão"** costuma obter a maior pontuação por ser o desfecho mais provável na linguagem cotidiana para essa frase física.

💡 **Para pensar:** Por que o modelo avalia a palavra "corpo" de forma genérica? Se mudássemos o contexto para "cair no *mar*" ou "cair no *rio*", o modelo se ajustaria? Veja como as sugestões refletem leis físicas e padrões comuns encontrados na internet onde o modelo estudou.

---

## Passo 4: Processando Múltiplas Sentenças em Lote (*Batch Processing*)
Na vida real, raramente processamos uma única frase por vez. Se tivermos milhares de textos para analisar, enviar um de cada vez tornaria o sistema lento. Os modelos de Machine Learning são otimizados para trabalhar com listas de dados de uma só vez.
```

````python
texto = mascarar(["Existe uma chance do corpo cair no [MASK]", "João Pessoa é a capital da [MASK]"])
for x in range(len(texto)):
  print(texto[x])
````

```markdown
### O que este passo faz e por que é necessário?
Em vez de passar apenas uma string simples para o nosso objeto `mascarar`, enviamos uma **lista de strings** `["Frase 1...", "Frase 2..."]`.
Isso permite que a placa de vídeo (GPU) ou o processador processe as sentenças em paralelo, economizando tempo computacional.

### Como interpretar o resultado?
A saída agora é uma lista de listas. Para cada frase enviada, recebemos um conjunto próprio de 5 sugestões de preenchimento.
* Na primeira frase ("Existe uma chance do corpo cair no..."), o modelo foca em conceitos físicos como "chão", "mar" ou "rio".
* Na segunda frase ("João Pessoa é a capital da..."), o modelo utiliza seu conhecimento enciclopédico e factual para sugerir com altíssima confiança (score próximo de 0.78) a palavra **"Paraíba"**, demonstrando que o modelo compreende relações geográficas reais presentes em seus dados de treinamento.

💡 **Para pensar:** Observe que na segunda frase, o modelo também sugeriu palavras de baixíssima probabilidade como "música" ou "bola". Isso ocorre porque, estruturalmente, a frase continua correta gramaticalmente ("João Pessoa é a capital da música"), mesmo que geograficamente a capital oficial seja a Paraíba. O modelo avalia tanto a gramática quanto o contexto semântico!

---

## 🚨 Erros Comuns e Dicas

1. **Erro de Token de Máscara Incorreto:**
   * **O que acontece:** O modelo `neuralmind/bert-base-portuguese-cased` foi treinado utilizando exatamente o marcador `[MASK]` (com colchetes e em letras maiúsculas). Se você tentar usar `<mask>` ou `_` (sublinhado), o modelo não reconhecerá a lacuna e gerará um erro informando que não encontrou o token de máscara.
   * **Dica:** Sempre verifique a documentação do modelo escolhido na Hugging Face para saber qual é o seu token de máscara padrão.

2. **Uso de Letras Maiúsculas/Minúsculas (Cased vs Uncased):**
   * **O que acontece:** Este modelo é "cased" (sensível a maiúsculas/minúsculas). Isso significa que ele diferencia "joão" de "João".
   * **Dica:** Ao testar entidades como nomes próprios ou capitais, certifique-se de escrever a frase respeitando a capitalização correta para obter melhores resultados.
```
