```markdown
## 🤖 Carregando o Modelo e Realizando a Transcrição

### O que este passo faz?
Carrega o cérebro da nossa IA (o modelo `base` do Whisper) na memória e solicita que ele transcreva o arquivo de áudio chamado `teste.mp3`.

### Por que ele é necessário?
1. **Modelos de IA (Machine Learning)** são redes neurais complexas. Nós usamos um modelo pré-treinado (que já aprendeu a ouvir e escrever) para não precisarmos treinar uma IA do zero.
2. Escolhemos o modelo `base`, que possui cerca de 74 milhões de parâmetros. Ele é um excelente equilíbrio entre velocidade e precisão para testes rápidos.
3. O método `.transcribe()` faz toda a mágica: detecta o idioma falado, separa o áudio em pequenos segmentos e gera o texto correspondente.

💡 **Para pensar**:
O código atualmente assume que o arquivo `teste.mp3` já está presente na pasta local do Colab. Se o arquivo não existir, o programa falhará com um erro de arquivo não encontrado (`FileNotFoundError`). Uma boa prática seria adicionar uma verificação prévia ou uma instrução de upload.
```

````python
modelo = whisper.load_model("base")
resultado = modelo.transcribe("teste.mp3")

texto = resultado['text']
idioma = resultado['language']

print(f'Transcrição: {texto} \nIdioma: {idioma}')
````

```markdown
## 💾 Salvando o Resultado em um Arquivo de Texto

### O que este passo faz?
Cria ou abre um arquivo físico chamado `arquivo.txt` no ambiente do Colab e escreve nele o conteúdo da variável `texto` (a transcrição obtida).

### Por que ele é necessário?
Quando o Google Colab é encerrado, todas as variáveis temporárias da memória RAM são apagadas. Salvar o resultado em um arquivo garante que você possa baixar sua transcrição e utilizá-la em outros programas ou enviá-la para alguém.
```

````python
with open('arquivo.txt', 'w') as f:
  f.write(f'{texto}')
````

1. **`FileNotFoundError: [Errno 2] No such file or directory: 'teste.mp3'`**

2. **Demora excessiva para rodar**
   * **Causa:** Processar áudio e rodar redes neurais pesadas na CPU pode ser demorado.
