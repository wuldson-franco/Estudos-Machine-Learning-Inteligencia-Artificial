```markdown
# Redução de Dimensionalidade com PCA na Prática 🧪

Este notebook apresenta o conceito e a aplicação prática da **Análise de Componentes Principais (PCA)** utilizando a base de dados Iris. Você aprenderá como reduzir o número de atributos de um conjunto de dados e o impacto que isso causa na acurácia de um modelo de classificação (neste caso, uma Floresta Aleatória).

## Objetivos de Aprendizagem 🎯
* Compreender o que é e por que utilizamos a **Redução de Dimensionalidade**.
* Aprender a padronizar os dados antes de aplicar o PCA.
* Criar modelos de Machine Learning (Random Forest) antes e depois da redução de dimensionalidade.
* Comparar os impactos na acurácia do modelo ao reduzir o número de variáveis.
```

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
````

```markdown
## Passo 1: Importação de Bibliotecas 📦

### O que este passo faz?
Ele carrega todas as ferramentas e algoritmos que utilizaremos ao longo do código.

### Por que ele é necessário?
O Python básico não possui funções nativas para Machine Learning complexo ou manipulação avançada de dados. Usamos a biblioteca `scikit-learn` (abreviada como `sklearn`), que é o padrão da indústria para ML em Python.

* **`train_test_split`**: Divide nossa base de dados entre uma parte para treino e outra para teste.
* **`accuracy_score`**: Uma métrica para avaliar o quão bom é o nosso classificador (percentual de acertos).
* **`PCA`**: O algoritmo de Redução de Dimensionalidade que vamos estudar.
* **`RandomForestClassifier`**: Um algoritmo de aprendizado supervisionado baseado em várias árvores de decisão (uma "floresta").
* **`StandardScaler`**: Ferramenta essencial para colocar todas as nossas variáveis na mesma escala numérica.
* **`datasets`**: Módulo que contém conjuntos de dados prontos para estudo, como o Iris.
```

````python
iris = datasets.load_iris()
previsores = iris.data
classe = iris.target
previsores
````

```markdown
## Passo 2: Carregando os Dados (Iris Dataset) 🌸

### O que este passo faz?
Carrega a clássica base de dados *Iris*, separando as características físicas das flores (como comprimento e largura da pétala e sépala) dos seus nomes científicos (as classes).

### Por que ele é necessário?
Para ensinar um modelo de Machine Learning, precisamos de dados históricos de exemplo.
* **`previsores`** (variáveis independentes): São as colunas com dados numéricos das características físicas.
* **`classe`** (variável dependente ou target): O resultado que queremos prever (o tipo de flor Iris: *Setosa*, *Versicolor* ou *Virginica*).

💡 **Para pensar**: O código carrega a base de dados Iris diretamente da biblioteca. Embora prático para estudo, na vida real você geralmente precisará ler os dados de arquivos externos (como arquivos CSV ou bancos de dados) usando bibliotecas como o Pandas.
```

````python
sc = StandardScaler()
previsores = sc.fit_transform(previsores)
previsores
````

```markdown
## Passo 3: Padronização dos Dados (Feature Scaling) ⚖️

### O que este passo faz?
O `StandardScaler` transforma as variáveis numéricas para que todas tenham média igual a 0 e desvio padrão igual a 1.

### Por que ele é necessário?
O PCA calcula a direção de máxima variância nos dados com base na distância geométrica dos pontos. Se uma variável tiver números muito maiores que as outras (por exemplo, salário em milhares vs. idade em dezenas), o PCA vai achar que essa variável é mais importante apenas porque os números dela são maiores. A **padronização** garante que todas as variáveis tenham o mesmo peso inicial.

### Como interpretar o resultado?
Ao inspecionar a matriz de previsores gerada, você notará valores decimais positivos e negativos bem pequenos (geralmente entre -3 e 3). Isso mostra que os dados foram centralizados com sucesso.
```

````python
previsores
````

```markdown
## Passo 4: Divisão em Treino e Teste (Sem PCA) ✂️

### O que este passo faz?
Divide os nossos previsores padronizados e a variável classe em quatro partes: dados de treino (`X_treinamento`, `y_treinamento`) e dados de teste (`X_teste`, `y_teste`). Usamos 30% dos dados para testes (`test_size=0.3`).

### Por que ele é necessário?
Não podemos avaliar um modelo usando os mesmos dados que ele usou para aprender, pois isso seria o equivalente a dar a um aluno uma prova com as mesmas perguntas que estavam no gabarito do material de estudo. A divisão garante uma avaliação honesta de dados novos.

* **`random_state=123`**: Define uma semente aleatória. Isso garante que a divisão seja idêntica toda vez que você rodar o notebook, tornando seus resultados reprodutíveis.
```

````python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(previsores, classe, test_size=0.3, random_state=123)
````

```markdown
## Passo 5: Treinamento e Avaliação da Floresta Aleatória (Dados Originais) 🌲

### O que este passo faz?
Instancia um classificador Random Forest com 100 árvores de decisão, realiza o treinamento (`fit`), faz previsões nos dados de teste e calcula a acurácia.

### Por que ele é necessário?
Este passo serve como a nossa **linha de base (baseline)**. Ele nos mostra quão bem o modelo performa utilizando todas as 4 dimensões (características) originais dos dados padronizados.

### Como interpretar o resultado?
O resultado de `taxa_acerto` é de aproximadamente **0.9333 (93.33% de acerto)**. Este é o número a ser batido ou comparado após a redução de dados.
```

````python
floresta = RandomForestClassifier(n_estimators=100, random_state = 1234)
floresta.fit(X_treinamento,y_treinamento)
previsoes = floresta.predict(X_teste)
taxa_acerto = accuracy_score(y_teste,previsoes)
taxa_acerto
````

```markdown
## Passo 6: Redução de Dimensionalidade com PCA 🧩

### O que este passo faz?
Instancia o algoritmo **PCA** para reduzir nossas 4 colunas originais para apenas 3 novos atributos (chamados de *Componentes Principais*). O `fit_transform` calcula esses novos eixos geométricos e projeta os dados neles.

### Por que ele é necessário?
**Definição técnica**: O PCA (Análise de Componentes Principais) é um método que projeta dados multidimensionais em eixos ortogonais que capturam a maior variação estatística possível do dataset original.

Em termos simples, se o nosso modelo de ML gasta muito processamento lidando com dezenas ou centenas de colunas, o PCA resume essa informação criando poucas colunas sintéticas novas que carregam a maior parte da informação original.

💡 **Para pensar**: O dataset Iris original só possui 4 colunas. Reduzir de 4 para 3 colunas não traz vantagens reais de performance computacional nesse cenário pequeno. O uso de PCA se mostra poderoso de verdade em bases de dados com dezenas, centenas ou milhares de características (como análise de imagens ou genética).
```

````python
pca = PCA(n_components=3)
previsores = pca.fit_transform(previsores)
previsores
````

```markdown
## Passo 7: Nova Divisão de Treino e Teste (Com PCA) ✂️

### O que este passo faz?
Faz novamente a separação de treino e teste, mas agora utilizando a variável `previsores` modificada, que agora possui apenas as 3 colunas sintetizadas pelo PCA.
```

````python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(previsores, classe, test_size=0.3, random_state=123)
````

```markdown
## Passo 8: Treinamento e Avaliação da Floresta Aleatória (Com PCA) 🌲🧪

### O que este passo faz?
Cria um novo modelo Random Forest idêntico ao anterior e o avalia com os dados reduzidos pelo PCA.

### Como interpretar o resultado?
A acurácia obtida foi de **0.9555 (95.55% de acerto)**.
Isso demonstra algo surpreendente: mesmo jogando fora uma parte dos dados (reduzindo de 4 dimensões para 3), a nossa acurácia aumentou ligeiramente! Isso acontece porque o PCA ajuda a eliminar "ruídos" estatísticos ou correlações redundantes presentes na base original.
```

````python
floresta = RandomForestClassifier(n_estimators=100, random_state = 1234)
floresta.fit(X_treinamento,y_treinamento)
previsoes = floresta.predict(X_teste)
taxa_acerto = accuracy_score(y_teste,previsoes)
taxa_acerto
````

```markdown
---

## ⚠️ Erros Comuns e Dicas de Sobrevivência
1. **Esquecer de Padronizar os Dados**: Aplicar PCA diretamente em variáveis com escalas muito diferentes fará com que o algoritmo dê preferência apenas à maior escala, destruindo a utilidade da redução de dimensões. Sempre use `StandardScaler` antes do PCA.
2. **Vazamento de Dados (Data Leakage)**: No mundo real, ao treinar modelos, idealmente aplicamos o `.fit()` do `StandardScaler` e do `PCA` apenas no conjunto de treino, usando apenas o `.transform()` no conjunto de teste. Aplicar em toda a base junta antes do split (como feito neste exemplo pedagógico) pode fazer com que informações de teste influenciem o treinamento.
3. **Perda de Interpretabilidade**: As colunas geradas pelo PCA (Componente 1, Componente 2, etc.) são combinações matemáticas das colunas originais. Você perde a capacidade de dizer diretamente "esta flor tem pétalas grandes", o que dificulta explicar o modelo para pessoas de negócios.

---

## Resumo Final 📝
* O **PCA** serve para reduzir o volume de características dos dados e acelerar o processamento eliminando redundâncias.
* A **Padronização** dos dados é obrigatória antes de se iniciar qualquer processo de PCA.
* Vimos na prática que a redução de dimensionalidade pode reter ou até mesmo melhorar o poder preditivo do modelo, removendo ruídos irrelevantes dos dados.
```
