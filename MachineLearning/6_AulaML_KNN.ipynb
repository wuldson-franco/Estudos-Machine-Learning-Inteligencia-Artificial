```markdown
# Aula Prática: Algoritmo KNN (K-Nearest Neighbors)

Este notebook apresenta uma introdução prática ao algoritmo **KNN (K-Nearest Neighbors)**, um dos métodos mais intuitivos e fundamentais de Aprendizado de Máquina Supervisionado. Aqui, utilizaremos o clássico conjunto de dados `mtcars` para classificar o número de cilindros de um veículo com base em suas características de consumo (mpg) e potência (hp).

## 🎯 Objetivos de Aprendizagem
Ao final deste notebook, você será capaz de:
1. Entender o conceito de **classificação supervisionada** e o funcionamento básico do algoritmo KNN.
2. Preparar dados, treinar o modelo utilizando a biblioteca `scikit-learn` e fazer previsões para novos dados.
3. Avaliar a performance de um classificador usando métricas como **Acurácia**, **Precisão**, **Recall** e **F1-Score**.
4. Compreender como a distância matemática (vizinhança) é utilizada para tomar decisões de classificação.

## 🛠️ Pré-requisitos e Bibliotecas Utilizadas
Para acompanhar este tutorial, é importante conhecer os conceitos básicos de Python (como listas, variáveis e importação de bibliotecas) e manipulação de dados com Pandas. Usaremos as seguintes bibliotecas:
* **`pandas` (pd):** Fundamental para carregar, visualizar e manipular tabelas de dados (DataFrames).
* **`numpy` (np):** Usada para manipulação de arrays multidimensionais e operações matemáticas de alta performance.
* **`sklearn` (scikit-learn):** A principal biblioteca de Machine Learning em Python. Dela, importaremos o modelo KNN (`KNeighborsClassifier`) e as funções de métricas para avaliar nosso modelo.
```

````python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import pandas as pd
import numpy as np
````

```markdown
## Passo 1: Carregamento dos Dados

Nesta etapa, importamos a base de dados que servirá de exemplo. O arquivo `mt_cars.csv` contém especificações técnicas de diversos modelos de carros.

* **Por que é necessário?** Todo projeto de Machine Learning começa com os dados. Precisamos carregar o arquivo para a memória do Python para conseguir analisá-lo e passá-lo para o modelo.
* **Como interpretar o resultado?** A função `.head()` exibe as 5 primeiras linhas da tabela (DataFrame). Isso nos ajuda a validar se o arquivo foi carregado corretamente e a identificar quais colunas temos disponíveis (como `mpg` - milhas por galão, `hp` - cavalos de potência, e `cyl` - cilindros).
```

````python
mtcars = pd.read_csv('mt_cars.csv')
mtcars.head()
````

```markdown
## Passo 2: Definição de Variáveis Preditoras (X) e Alvo (y)

Em problemas de aprendizado supervisionado, dividimos nossas variáveis em duas categorias:
* **Variáveis Preditoras (X):** As características (features) que usamos para fazer a previsão. No nosso caso, o consumo (`mpg`) e a potência (`hp`).
* **Variável Alvo (y):** O que queremos prever (target). Aqui, queremos descobrir a quantidade de cilindros (`cyl`) do carro.

* **Por que é necessário?** Os algoritmos do `scikit-learn` exigem essa separação clara entre as entradas (atributos de entrada) e as saídas (rótulo que queremos aprender a prever).
* **Como interpretar o resultado?** Criamos arrays NumPy chamados `X` e `y` contendo apenas os valores numéricos prontos para processamento.

> 💡 **Para pensar:** Repare que as variáveis `mpg` (que variam geralmente entre 10 e 35) e `hp` (que variam de 50 a mais de 300) estão em escalas muito diferentes. O algoritmo KNN baseia-se em distâncias matemáticas. O que acontece com a distância se uma variável tiver números muito maiores que a outra?
```

````python
X = mtcars[['mpg','hp']].values
y = mtcars['cyl'].values
````

```markdown
## Passo 3: Criação e Treinamento do Modelo KNN

O KNN significa **K-Vizinhos Mais Próximos**. Ele classifica um novo dado olhando para as `K` amostras mais parecidas (mais próximas fisicamente no gráfico) e realizando uma votação de maioria.

* **Por que é necessário?** Definimos `n_neighbors=3`, o que significa que o algoritmo olhará para os 3 carros mais parecidos com o carro que queremos classificar para decidir sua quantidade de cilindros. O método `.fit(X, y)` é onde o "treinamento" de fato acontece (no caso do KNN, ele apenas memoriza os pontos no espaço).
* **Como interpretar o resultado?** A saída `KNeighborsClassifier(n_neighbors=3)` indica que o objeto do modelo foi instanciado e treinado com sucesso com o parâmetro K=3.
```

````python
knn = KNeighborsClassifier(n_neighbors=3)
modelo = knn.fit(X,y)
modelo
````

```markdown
## Passo 4: Realizando Previsões na Base de Treino

Agora que o modelo está treinado, nós pedimos para ele tentar prever o número de cilindros de todos os carros que ele já conhecia na nossa base original `X`.

* **Por que é necessário?** Fazer previsões na própria base de treino nos permite avaliar o quão bem o modelo conseguiu aprender o padrão dos dados fornecidos.
* **Como interpretar o resultado?** O resultado é um array com as classes previstas para cada carro (por exemplo, `6`, `4` ou `8` cilindros). Cada número representa a previsão do modelo para o respectivo veículo.
```

````python
y_prev = modelo.predict(X)
print(y_prev)
````

```markdown
## Passo 5: Avaliação do Modelo

Para saber se o modelo é bom, precisamos comparar as previsões dele (`y_prev`) com os valores reais (`y`). Para isso, usamos quatro métricas importantes:
1. **Acurácia (Accuracy):** Proporção geral de acertos do modelo.
2. **Precisão (Precision):** Dentre os carros que o modelo previu que tinham X cilindros, quantos realmente tinham?
3. **Recall (Sensibilidade):** De todos os carros que realmente tinham X cilindros, quantos o modelo conseguiu identificar?
4. **F1-Score:** Uma média harmônica entre Precisão e Recall, excelente para entender o desempenho geral.
5. **Matriz de Confusão:** Uma tabela que mostra onde o modelo acertou e onde ele confundiu as classes.

* **Como interpretar os resultados?**
    * Uma **Acurácia de ~0.9375** significa que o modelo acertou aproximadamente 93.75% das classificações.
    * Na **Matriz de Confusão**, as linhas representam as classes reais (4, 6 e 8 cilindros) e as colunas as previsões. Os valores na diagonal principal (10, 6, 14) são os acertos. Os valores fora da diagonal são os erros (por exemplo, 1 carro de 4 cilindros foi classificado incorretamente como 6).
```

````python
accuracy = accuracy_score(y, y_prev)
precision = precision_score(y, y_prev, average='weighted')
recall = recall_score(y, y_prev, average='weighted')
f1 = f1_score(y, y_prev, average='weighted')
cm = confusion_matrix(y, y_prev)

print(f'Acurácia: {accuracy}, Precisão: {precision}, Recall: {recall}, F1: {f1}')
print('Matriz de confusão:\n', cm)
````

```markdown
## Passo 6: Classificando um Novo Carro

O objetivo final de um modelo é classificar dados novos que ele nunca viu antes. Criamos um carro fictício com `mpg = 19.3` e `hp = 105`.

* **Por que usar `.kneighbors()`?** Esta função nos retorna a distância e o índice (posição na tabela original) dos 3 vizinhos mais próximos desse novo carro.
* **Como interpretar o resultado?**
    * O modelo previu que o novo carro possui **6 cilindros**.
    * O método `.kneighbors()` nos revelou que os 3 vizinhos mais próximos estão localizados nos índices `5`, `31` e `1` da base original.
```

````python
#'mpg','hp'
new_data = np.array([[19.3,105]])

previsao = modelo.predict(new_data)
print(previsao)


distances, indices = modelo.kneighbors(new_data)
print(distances)
print(indices)
````

```markdown
## Passo 7: Inspecionando os Vizinhos Encontrados

Para validar a decisão do KNN, podemos olhar diretamente na tabela quais são esses carros de índices `1`, `5` e `31` que o algoritmo considerou mais próximos.

* **Como interpretar o resultado?**
    * Vemos que o carro no índice 1 tem 6 cilindros, o carro no índice 5 tem 6 cilindros e o carro no índice 31 tem 4 cilindros.
    * Como 2 dos 3 vizinhos mais próximos possuem **6 cilindros**, por voto de maioria, o algoritmo classificou corretamente o novo carro com 6 cilindros!
```

````python
mtcars.loc[[1,5,31],["cyl","mpg","hp"]]
````
