````python
# criação de classe para correlacionar os parametros
class LinearRegression:
    def __init__(self, x, y):
      self.x = x
      self.y = y
      self.__correlation_coefficient = self.__correlacao()
      self.__inclination = self.__inclinacao()
      self.__interception = self.__interceptacao()

    def __correlacao(self):
      covariacao = cov(self.x, self.y, bias=True)[0][1]
      variancia_x = var(self.x)
      variancia_y = var(self.y)
      return covariacao / sqrt(variancia_x * variancia_y)

    def __inclinacao(self):
      stdx = std(self.x)
      stdy = std(self.y)
      return self.__correlation_coefficient * (stdy / stdx)

    def __interceptacao(self):
      mediax = mean(self.x)
      mediay = mean(self.y)
      return mediay - mediax * self.__inclination

    def previsao(self, valor):
      return self.__interception + (self.__inclination * valor)
````

```markdown
## Passo 2: Criando a Estrutura da nossa Regressão Linear do Zero

### O que este passo faz?
Define a classe `LinearRegression` que encapsula toda a lógica matemática da Regressão Linear Simples.

### Por que ele é necessário?
A Regressão Linear busca encontrar a "linha perfeita" que descreve a relação entre uma variável preditora $X$ (entrada) e uma variável alvo $Y$ (saída). Essa linha é representada pela clássica equação da reta da geometria analítica:

$$Y = aX + b$$

Onde:
*   $a$ é a **Inclinação** (Slope): define o quão inclinada é a reta. Indica o quanto $Y$ muda para cada unidade de mudança em $X$.
*   $b$ é a **Interceptação** (Intercept): é o ponto onde a reta cruza o eixo vertical $Y$ (quando $X=0$).

Dentro da classe:
1.  `__correlacao()`: Calcula o **Coeficiente de Correlação** ($r$), que mede a força e a direção da relação linear entre $X$ e $Y$ (varia entre -1 e 1).
2.  `__inclinacao()`: Calcula a inclinação da reta usando a relação entre os desvios padrão de $Y$ e $X$ ponderada pelo coeficiente de correlação.
3.  `__interceptacao()`: Ajusta a altura da reta usando as médias de $X$ e $Y$.
4.  `previsao()`: Recebe um novo valor de entrada $X$ e calcula a estimativa de $Y$ correspondente.

💡 **Para pensar (Feedback do Professor):**
Repare que na linha `covariacao = cov(self.x, self.y, bias=True)[0][1]`, o parâmetro `bias=True` é utilizado para calcular a covariância populacional (dividida por $N$ em vez de $N-1$). É um detalhe estatístico crucial para garantir o alinhamento correto com os denominadores da variância calculada em seguida!
```

````python
# Criando uma arry para testar a classe
x = array([1,2,3,4,5])
y = array([2,4,6,8,10])

lr = LinearRegression(x,y)
previsao = lr.previsao(6)
print(previsao)
````
